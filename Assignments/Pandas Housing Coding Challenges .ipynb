## Task 14 Fill missing values with the median 

Topic: Missing data     

1. Start again from the original dataset and make a copy called filled.

3. Replace missing total_bedrooms values with that median. 

4. Print the median, the missing counts before and after, and the final row count.

Evidence: your code, the requested output and one sentence explaining the result. 

Complete your response in your notebook or submission document. 

Expected output

```python
filled = df.copy()

median_value = filled["total_bedrooms"].median()

filled["total_bedrooms"] = filled["total_bedrooms"].fillna(median_value)

print("Missing before:", df["total_bedrooms"].isna().sum())
print("Median value:", median_value)
print("Missing after:", filled["total_bedrooms"].isna().sum())
print("Rows after:", len(filled))
```

The code replaces missing values in total_bedrooms with the column's median value of 435.0, removing all 207 missing values without modifying the original DataFrame.

### Step by step explanation 

1. Start with a copy of the original df so the 207 missing entries are still present. 

2. median() calculates the middle value from the available numeric values, ignoring missing entries. 

3. The median for this file is 435.0. 

4. fillna() replaces missing entries only; existing bedroom values remain unchanged. 

5. Assign the result back to the column, then check that no missing values remain and all rows are retained. 

Median filling is an estimate, not a recovery of the original values. For predictive modelling, learn the median from training data only.
