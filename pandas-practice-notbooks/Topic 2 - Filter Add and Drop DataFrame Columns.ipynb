### Filter columns with bracket notation
- Let's consider the DataFrame from the previous topic

```python
np.random.seed(101)
df = pd.DataFrame(data=np.random.randn(5,4),
                  index=['A', 'B', 'C', 'D', 'E'],  
                  columns=['Col1', 'Col2', 'Col3', 'Col4']
                  )
df
```

- There are multiple methods to grab/select data from a DataFrame
  - Similarly to a NumPy array, you can select using brackets
  - Use one pair of brackets and pick the column you are interested in.

```python
df['Col1']
```

- The result is a pandas Series. You can check using type()

```python
type(df['Col1'])
```

You can select the same column and make the result a DataFrame. For that, use 2 pairs of brackets

  - The inner bracket is a list of the columns you are interested in, in this case, only one column

The result is a DataFrame. Most often, when subsetting data, you want the output as a DataFrame

You can grab more than one column

  - Add in a list of the columns you are interested in and place them inside the bracket

```python
df[['Col1','Col4']]
```

### Filter with .filter()¶

Another way to subset data in a Dataframe in a more pythonic way is with .filter(). Just parse in a list with the columns you are interested in using.

 - The example below retrieves Col1

```python
df.filter(['Col1'])
```

The example below retrieves 2 columns: Col1 and Col3

- The columns are parsed in a list
- The argument is axis=1, indicating you are grabbing columns

```python
df.filter(['Col1','Col3'],axis=1)
```

**PRACTICE:** grab the first 2 columns from the DataFrame df using the approach you just learned with .filter()

Your output should look like



```python
# write your code here
df.filter(["Col1", "Col2"], axis=1)
```

### Add Columns¶

image.png


You can do arithmetic operations and add new columns to your DataFrame.

Imagine if you need to sum Col1 and Col3, and the result should be stored at Col5
The code below says: create a Pandas Series named Col5. Get a series of Col1 and sum to a series of Col3. The Col5 series will be added to the DataFrame df.

```python
df['Col5'] = df['Col1'] + df['Col3']
df
```

### Drop Columns¶




You can drop columns or rows from your DataFrame, using .drop()

 - You set labels as your columns or index; axis=1 indicates you want to drop columns, axis=0, rows.
   - The example below drops Col5

```python
df.drop(labels=['Col5'],axis=1)
```

Check again df content

```python
df
```

You will notice Col5 is still there.

- There are 2 ways to solve this: using inplace=True or adjusting the code
- If you add inplace=True to df.drop(labels=['Col5'],axis=1), then you are overwriting the existing DataFrame

```python
df.drop(labels=['Col5'],axis=1,inplace=True)
df
```

Second approach: You can write df = df.drop(labels=['Col5'],axis=1), in this way you reassign the content of df.

 - But first, let's re-add Col5

```python
df['Col5'] = df['Col1'] + df['Col3']
df
```

**PRACTICE:** Remove Col5 with the second approach

Your df content should look like




```python
# write the code here using the second approach
df = df.drop(labels=['Col5'],axis=1)
df
```

### Drop Rows¶

You can drop rows using index reference, setting axis=0

- The use case is similar, using labels you parse in a list of the indices you want to drop

```python
df.drop(labels=['A','E'],axis=0)
```

Remember the example above doesn't remove the rows "for real" since the inplace argument is not set as True and the DataFrame was not reassigned

```python
df
```

**PRACTICE:** drop rows B and C from df, do not use the inplace argument

Your output should look like



```python
# write your code here
df.drop(labels=["B", "C"], axis=0, inplace=True)
df
```
